```python
from pyspark.sql import Row

# --------------------------------------------------
# 1. Generate sample dataset
# --------------------------------------------------

data = [
    (1, "Bhushan", "Data Engineer", 85000),
    (2, "Rahul", "Developer", 75000),
    (3, "Priya", "Analyst", 65000),
    (4, "Amit", "Data Engineer", 90000),
    (5, "Sneha", "Developer", 72000)
]

columns = ["emp_id", "emp_name", "role", "salary"]

df = spark.createDataFrame(data, columns)

# Display generated dataset
display(df)


# --------------------------------------------------
# 2. Create / overwrite Delta table
# --------------------------------------------------

table_name = "bronze.bi.employee_delta"

df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(table_name)


# --------------------------------------------------
# 3. Read the Delta table
# --------------------------------------------------

df_delta = spark.table(table_name)

display(df_delta)
```
